**Objective:** Learn how to import real-world tabular data for machine learning workflows.
**Description:** Load a CSV dataset using Pandas and display its first and last records.

**Dataset Used:** Iris Dataset (provided locally as `iris_dataset.csv`)

```python
# Import the pandas library
import pandas as pd

# Define the file path (Make sure the CSV is in the same directory as your notebook)
file_path = 'iris_dataset.csv'

# 1. Load the CSV dataset using pd.read_csv()
try:
    df = pd.read_csv(file_path)
    print("Dataset loaded successfully!\n")
except FileNotFoundError:
    print(f"Error: The file {file_path} was not found.")

# 2. Display the first five records using head()
print("--- First 5 Records ---")
display(df.head())

# 3. Display the last five records using tail()
print("\n--- Last 5 Records ---")
display(df.tail())
```

### Interview Questions & Answers

**1. What is a CSV file?**
A CSV (Comma-Separated Values) file is a plain text file that stores tabular data (numbers and text) in plain text. Each line of the file is a data record, and each record consists of one or more fields, separated by commas. It is a very common and simple format used to exchange data between different applications and systems.

**2. How do you load a CSV file using Pandas?**
You can load a CSV file in Pandas using the `pd.read_csv()` function. You simply pass the file path (or URL) as an argument to the function. For example: `df = pd.read_csv('filename.csv')`. It automatically parses the tabular data into a Pandas DataFrame.

**3. What is a DataFrame?**
A DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure in Pandas. It is similar to a spreadsheet or SQL table, consisting of rows and columns. It is the primary data structure used in Pandas for data manipulation, analysis, and cleaning.
